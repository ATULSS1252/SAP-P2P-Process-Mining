# 01 — Load and explore the BPI Challenge 2019 OCEL

Loads the object-centric event log, sanity-checks it against the published dataset scale, and does a first pass over the object and event types before any process discovery happens in `02_bottleneck_analysis.ipynb`.

```python
import os
import pm4py

# Load the BPI Challenge 2019 OCEL (Object-Centric Event Log).
# See ../data/README.md for how to get this file — it is not committed here.
ocel_path = os.path.join("..", "data", "BPIC19.jsonocel")
print(f"Loading OCEL from {ocel_path} ...")
ocel = pm4py.read_ocel(ocel_path)

print("\n--- OCEL summary ---")
print(ocel)
```

```python
n_events = len(ocel.events)
n_objects = len(ocel.objects)
print(f"Events:  {n_events:,}")
print(f"Objects: {n_objects:,}")

# Sanity check against the published dataset scale:
# 76,349 purchase documents / 251,734 line items / 1,595,923 events (42 activities)
assert n_events > 1_000_000, "Event count looks too low — check the data file."
print("\nScale check passed.")
```

## Object and event type breakdown

```python
object_type_counts = ocel.objects["ocel:type"].value_counts()
activity_counts = ocel.events["ocel:activity"].value_counts()

print("Object types (count per type):")
print(object_type_counts)

print("\nTop 15 activities by event count:")
print(activity_counts.head(15))
```
